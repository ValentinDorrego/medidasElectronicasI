Se busca medir la impedancia de un dipolo paralelo conformado por un  capacitor y una resistencia cuya fase se encuentra entre 25º/40º a una frecuencia de 1kHz. Para ello se propone medir la tensión y corriente del dipolo utilizando un DSO y un resistencia en serie para medir la corriente.
Se debe tener cuidado al definir las resistencias para que las mediciones de  tensión esté en la misma escala y de no cortocircuitar las masas del DSO.

La impedancia compleja Z = V/I se obtiene midiendo módulo y fase relativa entre la tensión en el dipolo y la corriente que lo atraviesa. 

Como el DSO solo mide tensiones, la corriente se obtiene indirectamente con una resistencia en serie (shunt) Rs.
- CH1 mide la tensión total (V_Z + V_Rs), del generador a masa.
- CH2 mide V_Rs, entre el punto intermedio y masa.
-Luego V_Z = CH1 − CH2 (con la función MATH del DSO, restando fasorialmente).

Usar el DSO con CH1 − CH2 (modo diferencial/MATH) para obtener V_Z directamente.

$$

*Medición de fase*

El retardo se mide entre CH1 (tensión total del generador, referencia) y CH2 ($V_z$, tensión sobre el dipolo). CH2 está retrasado respecto de CH1, por lo que el ángulo es negativo.

$$
T = 1\,\text{ms} \qquad D_T = 100\,\mu\text{s}
$$

**Incertidumbre de T** (base de tiempo ±0,01 %):

$$
\Delta T = 0{,}01\%\cdot T = 0{,}0001\cdot 1\,\text{ms} = 0{,}1\,\mu\text{s}
$$

$$
u(T) = \frac{\Delta T}{\sqrt{3}} = 57{,}7\,\text{ns}
\qquad
u_R(T) = \frac{57{,}7\,\text{ns}}{1\,\text{ms}} = 5{,}77\cdot10^{-5}
$$

**Incertidumbre de $D_T$** (cursores ±2 % + base de tiempo ±0,01 %):

$$
\Delta D_T = (2\% + 0{,}01\%)\cdot D_T = 0{,}0201\cdot 100\,\mu\text{s} = 2{,}01\,\mu\text{s}
$$

$$
u(D_T) = \frac{\Delta D_T}{\sqrt{3}} = 1{,}16\,\mu\text{s}
\qquad
u_R(D_T) = \frac{1{,}16\,\mu\text{s}}{100\,\mu\text{s}} = 1{,}16\cdot10^{-2}
$$

**Fase:**

$$
|\theta_{V_z}| = \frac{D_T}{T}\cdot 360^\circ = \frac{100\,\mu\text{s}}{1\,\text{ms}}\cdot 360^\circ = 36^\circ
$$

$$
u_R(\theta_{V_z}) = \sqrt{u_R^2(T) + u_R^2(D_T)} = \sqrt{(5{,}77\cdot10^{-5})^2 + (1{,}16\cdot10^{-2})^2} \approx 1{,}16\cdot10^{-2}
$$

$$
u(\theta_{V_z}) = 36^\circ\cdot 1{,}16\cdot10^{-2} = 0{,}42^\circ
\qquad
U(\theta_{V_z}) = k_{95\%}\cdot u(\theta_{V_z}) = 2\cdot 0{,}42^\circ = 0{,}84^\circ
$$

$$
\boxed{\theta_{V_z} = -(36{,}0 \pm 0{,}8)^\circ \quad (k=2)}
$$

$$
\boxed{Z = (598 \pm 74)\,\Omega\ \angle\ -(36{,}0 \pm 0{,}8)^\circ}
