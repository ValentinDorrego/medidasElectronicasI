$$

- Incertidumbre tipo B
$$

$$

- Factor de cobertura utilizada

## Verificación del instrumento

Una vez calculados el valor verdadero ($\overline{V_p}$) y el valor medido ($\overline{V_m}$) para cada punto de calibración, se evalúa si el multímetro genérico (UNI-T UT61E) cumple con la exactitud declarada por el fabricante.

### 1. Desviación

Diferencia entre el valor medido por el instrumento a calibrar y el valor verdadero (patrón):

$$
\Delta v = \overline{V_m} - \overline{V_p}
$$

### 2. Incertidumbre expandida del instrumento a calibrar

Se obtiene multiplicando la incertidumbre combinada ($u_c$) por el factor de expansión ($k$), típicamente $k=2$ para un nivel de confianza del 95%:

$$
U(\overline{V_m}) = k \cdot u_c(\overline{V_m})
$$

### 3. Semirrango del fabricante ($a_m$)

Es la tolerancia que promete la hoja de datos del UT61E para la escala utilizada, evaluada en el valor medido. Se calcula con el porcentaje de lectura y los dígitos de exactitud que indica el datasheet:

$$
a_m = \left( \frac{\%_{fabricante}}{100} + \frac{n}{C_i} \right) \cdot \overline{V_m}
$$

Donde:
- $\%_{fabricante}$: porcentaje de exactitud declarado por el fabricante para esa escala
- $n$: cantidad de dígitos de incertidumbre declarados por el fabricante
- $C_i$: valor medido expresado en cuentas, sin coma decimal (ej: 100.44mV → $C_i = 10044$)

### 4. Evaluación de la condición

El instrumento **verifica** si la desviación, sumada a su propia incertidumbre expandida, entra dentro del semirrango que promete el fabricante:

$$
|\Delta v| + U(\overline{V_m}) \leq a_m
$$

- Si se cumple → **Verifica: Sí** ✅
- Si no se cumple → **Verifica: No** ❌

$$
